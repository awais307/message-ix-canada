```python
import pandas as pd

msg_provs = {
    "NEWFOUNDLAND AND LABRADOR": "NewfoundlandandLabrador",
    "NEW BRUNSWICK": "NewBrunswick",
    "ALBERTA": "Alberta",
    "BRITISH COLUMBIA": "BritishColumbia",
    "NOVA SCOTIA": "NovaScotia",
    "ONTARIO": "Ontario",
    "QUEBEC": "Quebec",
    "SASKATCHEWAN": "Saskatchewan",
    "YUKON": "Yukon",
    "NORTHWEST TERRITORIES":  "NorthwestTerritories",
    "PRINCE EDWARD ISLAND": "PrinceEdwardIsland",
    "MANITOBA": "Manitoba",
    "NUNAVUT":"Nunavut"
}

# Natural Gas: 10^6 m3 = 37300 GJ --> 37300 GJ = 10.36 GWh
converson_factor = 10.36

def process_gas_file(file):
    df = pd.read_csv(file)
    # Extract unique values from the original DataFrame
    provinces = list(df['region'].unique())
    #selection = list(df['selection'].unique())[0]
    scenario = list(df['scenario'].unique())[0]
    unit = list(df['unit'].unique())[0]
    #dataset = list(df['dataset'].unique())[0]
    years = list(df['year'].unique())

    rows = []

    for province in provinces:
        for year in [2020, 2025, 2030, 2035, 2040, 2045, 2050]:
            summed_value = 0

            value = df.loc[df['source'].isin(['CONVENTIONAL', 'SOLUTION']) & df['region'].isin([province]) & df['year'].isin([year]), 'value'].sum()
            
            #gas_extr_1 (40%), gas_extr_2 (25%), gas_extr_3 (20%), gas_extr_3 (15%)
            source = "gas_extr_1"
            row = [scenario, msg_provs[province], source, year, "M1", "year", value*0.4*converson_factor, "GWh"]
            rows.append(row)
            summed_value += value*0.4

            source = "gas_extr_2"
            row = [scenario, msg_provs[province], source, year, "M1", "year", value*0.25*converson_factor, "GWh"]
            rows.append(row)
            summed_value += value*0.25

            source = "gas_extr_3"
            row = [scenario, msg_provs[province], source, year, "M1", "year", value*0.2*converson_factor, "GWh"]
            rows.append(row)
            summed_value += value*0.2

            source = "gas_extr_4"
            row = [scenario, msg_provs[province], source, year, "M1", "year", value*0.15*converson_factor, "GWh"]
            rows.append(row)
            summed_value += value*0.15

            value = df.loc[df['source'].isin(['COAL BED METHANE', 'SHALE', 'TIGHT']) & df['region'].isin([province]) & df['year'].isin([year]), 'value'].sum()
            
            #gas_extr_5 (20%), gas_extr_6 (80%)
            source = "gas_extr_5"
            row = [scenario, msg_provs[province], source, year, "M1", "year", value*0.2*converson_factor, "GWh"]
            rows.append(row)
            summed_value += value*0.2

            source = "gas_extr_6"
            row = [scenario, msg_provs[province], source, year, "M1", "year", value*0.8*converson_factor, "GWh"]
            rows.append(row)
            summed_value += value*0.8

            source = "gas_extr_mpen"
            row = [scenario, msg_provs[province], source, year, "M1", "year", summed_value*converson_factor, "GWh"]
            rows.append(row)


    # Convert the list of rows into a DataFrame
    final_df = pd.DataFrame(rows, columns=['scenario','node_loc', 'technology', 'year_act', 'mode', 'time', 'value', 'unit'])
    #final_df = pd.DataFrame(rows, columns=['region', 'source', 'scenario', 'year', 'value', 'unit'])

    # Display the resulting DataFrame
    print(final_df.head()) 

    final_df.to_csv(f'gas_{scenario}.csv', index=False)
```

```python
process_gas_file('gas_current.csv')
process_gas_file('gas_canada_net_zero.csv')
process_gas_file('gas_global_net_zero.csv')
```
