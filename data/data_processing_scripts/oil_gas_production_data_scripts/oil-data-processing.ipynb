```python
import pandas as pd

def process_oil_file(file):

    df = pd.read_csv(file)

    source_map = {
        "C5+ (PENTANES PLUS)" : "oil_extr_2",
        "FIELD CONDENSATE" : "oil_extr_4",
        "CONVENTIONAL HEAVY" : "oil_extr_3",
        "IN SITU BITUMEN" : "oil_extr_5",
        "CONVENTIONAL LIGHT" : "oil_extr_1",
        "MINED BITUMEN" : "oil_extr_6"
    }

    sources = df["source"].unique()
    years = df["year"].unique()
    provs = df["region"].unique()

    data_dict = {
        int(year): {
            prov:{
                source_map[source]: float(df[(df["year"] == year) & (df["source"] == source) & (df['region'] == prov)]["value"].values[0])
                for source in sources
                if not df[(df["year"] == year) & (df["source"] == source)].empty
            } for prov in provs
        }
        for year in years
    }

    msg_provs = {
        "NEWFOUNDLAND AND LABRADOR": "NewfoundlandandLabrador",
        "NEW BRUNSWICK": "NewBrunswick",
        "ALBERTA": "Alberta",
        "BRITISH COLUMBIA": "BritishColumbia",
        "NOVA SCOTIA": "NovaScotia",
        "ONTARIO": "Ontario",
        "QUEBEC": "Quebec",
        "SASKATCHEWAN": "Saskatchewan",
        "YUKON": "Yukon",
        "NORTHWEST TERRITORIES":  "NorthwestTerritories",
        "PRINCE EDWARD ISLAND": "PrinceEdwardIsland",
        "MANITOBA": "Manitoba",
        "NUNAVUT":"Nunavut"
    }

    msg_years = {
        "2010":	[2006, 2007, 2008, 2009, 2010],
        "2015":	[2011, 2012, 2013, 2014, 2015],
        "2020":	[2016, 2017, 2018, 2019, 2020],
        "2025":	[2021, 2012, 2013, 2014, 2025],
        "2030":	[2026, 2027, 2028, 2029, 2030],
        "2035": [2031, 2032, 2033, 2034, 2035],
        "2040": [2036, 2037, 2038, 2039, 2040],
        "2045": [2041, 2042, 2043, 2044, 2045],
        "2050": [2046, 2047, 2048, 2049, 2050],
    }

    data_dict_averaged = {
        year: {
            prov:{
                source: sum([data_dict[msg_year][prov][source] for msg_year in msg_years[year]])/5 for source in list(source_map.values())
            } for prov in provs
        } for year in list(msg_years.keys())
    }

    # Initialize an empty list to store rows
    rows = []

    # Extract unique values from the original DataFrame
    provinces = list(df['region'].unique())
    scenario = list(df['scenario'].unique())[0]
    unit = list(df['unit'].unique())[0]

    # Oil (assuming it is heavy fuel oil): 1m3 = 41.73 GJ = 0.0116 GWh
    converson_factor = 0.0116

    # Loop through years and sources
    for year in msg_years.keys():
        for province in provinces:

            summed_value = 0

            for source in source_map.values():
                value = data_dict_averaged[year][province][source]  # Extract the value from the dictionary
                # Create a row
                row = [scenario, msg_provs[province], source, year, "M1", "year", value*converson_factor, "GWh"]

                # Append the row to the list
                rows.append(row)
                summed_value += value

            # Create a row
            row = [scenario, msg_provs[province], "oil_extr_mpen", year, "M1", "year", summed_value*converson_factor, "GWh"]
            # Append the row to the list
            rows.append(row)


    # node_loc,technology,year_act,mode,time,value,unit
    # Convert the list of rows into a DataFrame
    new_df = pd.DataFrame(rows, columns=['scenario','node_loc', 'technology', 'year_act', 'mode', 'time', 'value', 'unit'])

    # Display the resulting DataFrame
    print(new_df.head())

    new_df.to_csv(f'oil_{str.lower(scenario)}.csv', index=False)
```

```python
process_oil_file("oil_canada_net_zero.csv")
process_oil_file("oil_global_net_zero.csv")
process_oil_file("oil_current.csv")
```
