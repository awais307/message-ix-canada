```python
# open test_Alberta.xlsx and get all the sheet names
xls = pd.ExcelFile('test_Alberta.xlsx')
sheet_names = xls.sheet_names
dfs = []
for sheet in sheet_names:
    df = pd.read_excel('test_Alberta.xlsx', sheet_name=sheet)
    variable, unit = sheet.split(' (')
    # strip spaces from variable and unit left and right and remove parantheses
    variable = variable.strip()
    
    unit = unit.strip()
    unit = unit[:-1]
    df.rename(columns={df.columns[0]: 'time'}, inplace=True)

    # melt with all columns except time as id_vars and variable name as var_name
    df = pd.melt(df, id_vars=['time'], var_name='variable', value_name='value')
    df['unit'] = unit
    df['variable'] = variable + '|' + df['variable']
    dfs.append(df)

results = pd.concat(dfs)
```

```python
results = pd.concat(dfs)
```
