```python
import pandas as pd
```

```python
input_filename = 'MESSAGE_test_updated_LR.xlsx'
output_filename = 'MESSAGE_test_updated_LR.csv'
```

```python
import argparse
import os

def process_file(input_filename, output_filename):

    f_name, extension = os.path.splitext(input_filename)
    # Read Message data excel file (has only one sheet)
    df = pd.read_excel(input_filename, sheet_name=f_name)

    # Drop columns and rows that are all NaN
    df = df.dropna(axis=1).dropna(axis=0)

    # Melt the dataframe
    df_melted = df.melt(id_vars=['Model', 'Scenario', 'Region', 'Variable'], var_name='Time', value_name='Value')
    df_melted['Unit'] = 'Test'

    # Save the melted dataframe to a CSV file
    df_melted.to_csv(output_filename, index=False)
```

```python
process_file(input_filename,output_filename)
```
