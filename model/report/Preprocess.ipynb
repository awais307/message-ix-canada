```python
#Preprocess MESSAGEix-Canada Output before validation process
import pandas as pd
import numpy as np

# Load the Excel file
file_path = "MESSAGEix-CA_baseline 2.xlsx"  
df_correction = pd.read_excel(file_path)

# Replace empty values in the 'region' column with 'Unknown'
df_correction['Region'] = df_correction['Region'].replace(np.nan, 'Unkown')

df_correction.to_csv("MESSAGEix-CA_baseline 2_Region_Corrected.csv", index=False)

#check douplication
duplicates = df_correction[df_correction.duplicated()]

# Display the duplicated rows
print("Duplicated rows:")
print(duplicates)

num_duplicates = duplicates.shape[0]
print(f"Number of duplicated rows: {num_duplicates}")

# Remove the duplicates
df_correction = df_correction.drop_duplicates()

# Save the dataframe without duplicates as input for validation process
df_correction.to_csv("MESSAGEix-CA_baseline 2_Preprocessed.csv", index=False)
```
