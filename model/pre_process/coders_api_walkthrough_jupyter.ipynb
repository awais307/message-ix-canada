## Generators

```python
generators_df = pd.DataFrame.from_dict(requests.get(url+"/generators"+query).json())
```

```python
generators_df.columns.tolist()
```

```python
generators_df['capacity_adjustment'].sort_values()
```

```python
def print_generator_col_members(df, col_index=None, col_name=None):
    assert col_index or col_name
    if col_index:
        col_name = generators_df.columns[col_index]
    col_unique_members = generators_df[col_name].unique()
    col_unique_members.sort()
    print("Unique members of {} column ({} in total)".format(col_name, len(col_unique_members)))
    print(col_unique_members)
```

### Regions

#### Provinces

```python
print_generator_col_members(generators_df, col_name="province")
```

#### copper_balancing_area

```python
print_generator_col_members(generators_df, col_name="copper_balancing_area")
```

#### planning_region

```python
print_generator_col_members(generators_df, col_name="operating_region")
```

#### Start years

```python
print_generator_col_members(generators_df, col_name="start_year")
```

```python
print("Number of future projects: ", generators_df[generators_df["start_year"]>2021].shape[0])
print("Types of projects: ")
for gen_type, df in generators_df[generators_df["start_year"]>2021].groupby("gen_type"):
    print(gen_type, "{} projects".format(df.shape[0]), "Capacity: {} MW".format(df["unit_installed_capacity"].sum()))
```

#### End years

```python
print_generator_col_members(generators_df, col_name="closure_year")
```

```python
print("Number of projects with planned closures: ", generators_df[generators_df["closure_year"]<2022].shape[0])
print("Types of projects: ")
for gen_type, df in generators_df[generators_df["closure_year"]<2022].groupby("gen_type"):
    print(gen_type, "{} projects".format(df.shape[0]), "Capacity: {} MW".format(df["unit_installed_capacity"].sum()))
```

#### Life times

```python
generators_df["lifetime"]=generators_df["closure_year"]-generators_df["start_year"]
print_generator_col_members(generators_df, col_name="lifetime")
```

```python
print("Types of projects with different lifetimes: ")
lifetime_ranges = [(i*25, i*25+25) for i in range(6)]
for r in lifetime_ranges:
    df = generators_df.loc[
        (generators_df["lifetime"]<=r[1]) & (generators_df["lifetime"]>r[0]), :]    
    if df.shape[0]==0:
        continue
    print("Lifetime range: [{}, {}]".format(r[0], r[1]))
    for gen_type, df_ in df.groupby("gen_type"):
        print("\t", gen_type, "{} projects".format(df_.shape[0]), "Capacity: {} MW".format(df_["unit_installed_capacity"].sum()))
```
